```python
# K-Means Clustering and Elbow Method

# Step 1: Import libraries
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Step 2: Load the dataset
iris = load_iris()
X = iris.data[:, :2]

print("Dataset shape:", X.shape)

# Step 3: Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 4: Apply the Elbow Method
inertia = []

for k in range(1, 11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X_scaled)
    inertia.append(model.inertia_)

# Step 5: Plot the Elbow graph
plt.plot(range(1, 11), inertia, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.grid()
plt.show()

# Step 6: Apply K-Means with K = 3
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_scaled)

# Step 7: Visualize the clusters
plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=clusters,
    cmap='viridis',
    s=50
)

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker='X',
    s=200,
    color='red',
    label='Centroids'
)

plt.xlabel("Standardized Sepal Length")
plt.ylabel("Standardized Sepal Width")
plt.title("K-Means Clustering (K = 3)")
plt.legend()
plt.show()

# Step 8: Display cluster assignments
print("\nFirst 10 cluster assignments:")
print(clusters[:10])

# Step 9: Calculate inertia manually using an example

# Cluster 1
cluster1 = [2, 3, 4]
centroid1 = sum(cluster1) / len(cluster1)

inertia1 = sum((x - centroid1) ** 2 for x in cluster1)

print("\nCluster 1 Centroid:", centroid1)
print("Cluster 1 Inertia:", inertia1)

# Cluster 2
cluster2 = [7, 8, 9]
centroid2 = sum(cluster2) / len(cluster2)

inertia2 = sum((x - centroid2) ** 2 for x in cluster2)

print("\nCluster 2 Centroid:", centroid2)
print("Cluster 2 Inertia:", inertia2)

# Total inertia
total_inertia = inertia1 + inertia2

print("\nTotal Inertia:", total_inertia)
```
