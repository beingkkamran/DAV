```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA

data = {
    'Study_Hours': [2, 4, 6, 8, 10, 3, 5, 7, 9, 12],
    'Attendance': [60, 65, 70, 75, 80, 62, 68, 78, 85, 95],
    'Assignment_Score': [40, 50, 55, 65, 70, 45, 60, 72, 80, 90],
    'Exam_Score': [45, 50, 60, 65, 75, 48, 58, 70, 82, 95]
}

df = pd.DataFrame(data)

print("ORIGINAL DATASET:")
print(df)

print("\nSTATISTICAL SUMMARY:")
print(df.describe())

scaler_minmax = MinMaxScaler()
normalized_data = scaler_minmax.fit_transform(df)

normalized_df = pd.DataFrame(
    normalized_data,
    columns=df.columns
)

print("\nNORMALIZED DATASET:")
print(normalized_df.round(2))

print("\nMINIMUM VALUES AFTER NORMALIZATION:")
print(normalized_df.min().round(2))

print("\nMAXIMUM VALUES AFTER NORMALIZATION:")
print(normalized_df.max().round(2))

scaler_standard = StandardScaler()
standardized_data = scaler_standard.fit_transform(df)

standardized_df = pd.DataFrame(
    standardized_data,
    columns=df.columns
)

print("\nSTANDARDIZED DATASET:")
print(standardized_df.round(2))

print("\nMEAN OF EACH FEATURE:")
print(standardized_df.mean().round(2))

print("\nSTANDARD DEVIATION OF EACH FEATURE:")
print(standardized_df.std(ddof=0).round(2))

scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

pca = PCA(n_components=2)
pca_data = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame(
    pca_data,
    columns=['PC1', 'PC2']
)

print("\nPCA TRANSFORMED DATA:")
print(pca_df.round(2))

print("\nEXPLAINED VARIANCE RATIO:")
print(pca.explained_variance_ratio_.round(4))

print("\nTOTAL VARIANCE RETAINED:")
print(round(pca.explained_variance_ratio_.sum(), 4))

print("\nVARIANCE RETAINED IN PERCENTAGE:")
print(round(pca.explained_variance_ratio_.sum() * 100, 2), "%")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

normalized_df.boxplot(ax=axes[0])
axes[0].set_title("Normalized Features")
axes[0].set_ylabel("Scaled Values")
axes[0].tick_params(axis='x', rotation=45)

standardized_df.boxplot(ax=axes[1])
axes[1].set_title("Standardized Features")
axes[1].set_ylabel("Standardized Values")
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))

plt.scatter(
    pca_df['PC1'],
    pca_df['PC2'],
    s=60
)

for i in range(len(pca_df)):
    plt.annotate(
        f"Student {i + 1}",
        (pca_df['PC1'][i], pca_df['PC2'][i])
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA: Dimensionality Reduction")
plt.grid(True)
plt.tight_layout()
plt.show()

print("\nFEATURE ENGINEERING COMPLETED SUCCESSFULLY.")
```
